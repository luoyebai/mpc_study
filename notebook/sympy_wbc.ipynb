$$M(q)\ddot{q}+C(q, \dot{q})\dot{q} + g(q) = \tau+J_c^T\lambda$$

<center>

$q$ 是广义坐标（关节角、位姿等）<br>
$M(q)$ 是广义质量矩阵<br>
$C(q, \dot{q})$ 是科氏力矩矩阵<br>
$g(q)$ 是重力项<br>
$\tau$ 是关节力矩控制输入<br>
$J_c^T$ 是接触雅可比矩阵<br>
$\lambda$ 是接触力（用于接触约束的拉格朗日乘子）

WBC优化目标函数定义：
$$ 
\underset{\dot{q}, \tau, \lambda}{min} \frac{1}{2} \| A\ddot{q}-b\|^2 + \alpha \| tau \|^2
$$ 

$A\ddot{q}=b$表示任务目标，例如轨迹跟踪、姿态控制等<br>
$\alpha \| \tau \| ^ 2$ 是用于最小化关节力矩的正则化项，防止关节力矩过大<br>
$ A,b$ 根据任务具体定义，如关节角度轨迹、手部末端位置等<br>

</center>
